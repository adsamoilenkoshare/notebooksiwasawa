### 1. The Herbrand–Ribet theorem

Let $p$ be an odd prime, $F=\mathbb{Q}(\mu_p)$, $\Delta=\operatorname{Gal}(F/\mathbb{Q})\cong(\mathbb{Z}/p\mathbb{Z})^\times$, and let $A$ be the $p$-part of the ideal class group of $F$.

Let $\omega:\Delta\to\mathbb{Z}_p^\times$ be the Teichmüller character. Since $\#\Delta=p-1$ is prime to $p$, $A$ splits into eigenspaces

$$A=\bigoplus_{i \bmod (p-1)} A^{\omega^i},\qquad A^{\chi}=e_\chi A,\qquad e_\chi=\frac{1}{p-1}\sum_{\sigma\in\Delta}\chi^{-1}(\sigma)\,\sigma .$$

**Theorem (Herbrand 1932; Ribet 1976).** For odd $i$ with $3\le i\le p-2$:

$$A^{\omega^i}\neq 0 \iff p \mid B_{p-i} \left( \iff p\mid B_{1,\omega^{-i}}\right),$$

where $B_{1,\psi}=\frac1p\sum_{a=1}^{p-1}a\,\psi(a)$, and the last equivalence is the congruence $B_{1,\omega^{j}}\equiv B_{j+1}/(j+1)\pmod p$ for $j\not\equiv 0,-1 \pmod{p-1}$.

Herbrand's direction ($\Rightarrow$) is a corollary of Stickelberger's theorem. Part II develops an approach to Ribet's converse using cyclotomic units. Kolyvagin's Euler system of Gauss sums, in Rubin's exposition, reproves the converse and in fact gives the exact order $\#A^{\omega^i}=p^{v_p B_{1,\omega^{-i}}}$.
**Example $p=37$.** The only even $j\le 37$ with $37\mid B_j$ is $j=32$, so $(37,32)$ is the unique irregular pair, $i=p-j=5$ is the unique irregular odd index, and in fact $A=A^{\omega^5}\cong\mathbb{Z}/37\mathbb{Z}$ (class number of $\mathbb{Q}(\zeta_{37})$ is exactly $37$).

The following functions are included directly in the notebook. They compute
$$B_{1,\omega^e}=\frac1p\sum_{a=1}^{p-1}a\,\omega(a)^e\pmod M.$$
Extra $p$-adic precision is used before division by $p$. In particular, the example below calls `B1_OmegaPow(31, 37)`.

```gp
TeichLiftMod(a,p,alpha) =
{
    my(PA = p^alpha, u = lift(Mod(a,PA)));
    if(gcd(u,p) != 1, error("TeichLiftMod: a must be prime to p"));
    for(i=1, alpha+2, u = lift(Mod(u,PA)^p));
    return(u);
}

B1_OmegaPow(e,M) =
{
    my(s = valuation(M,p), prec = p^(s+2), tot);
    e = e % (p-1);
    if(e == 0,
        tot = p*(p-1)/2,
        tot = lift(sum(a=1, p-1, a*Mod(TeichLiftMod(a,p,s+2), prec)^e)));
    if(tot % p, error("B_{1,omega^e} is not p-integral (e = -1 mod p-1)"));
    return(Mod(tot/p, M));
}
```

### 2. Stickelberger's theorem

Let $\tau_a:\zeta\mapsto\zeta^a$, and define

$$\theta = \sum_{\substack{a\bmod p\\ (a,p)=1}} \Big\{\frac{a}{p}\Big\}\tau_a^{-1} \in \mathbb{Q}[\Delta].$$

Fix $c$ primitive root modulo $p$ (for $p=37$ choose $c=2$), and define

$$\theta(1)=(\tau_c-c)\theta$$

**Theorem (Stickelberger).** $\theta(1)$ annihilates the ideal class group of $\mathbb{Q}(\zeta_{p})$.

### 3. Herbrand's theorem as a corollary of Stickelberger's theorem

Denote by $\chi$ some $\mathbb{Z}_p^{\times}$-valued character of $\Delta$. (For $p=37$ let $\chi=\omega^5$).

Project $\theta(1)$ to the $\chi$-eigenspace for an odd $\chi=\omega^k\ne\omega$. A direct computation gives $$e_\chi\theta = B_{1,\chi^{-1}}\,e_\chi,$$ so $(\chi(c)-c)\,B_{1,\chi^{-1}}$ annihilates $A^{\chi}$. Choosing $c$ a primitive root makes $c-\chi(c)\equiv c-c^{k}\not\equiv0\pmod p$ (as $k\neq1$), a unit. Hence

$$B_{1,\chi^{-1}}\ \text{annihilates}\ A^{\chi}
\qquad\Longrightarrow\qquad
A^{\omega^i}\neq0 \Rightarrow p\mid B_{1,\omega^{-i}} \equiv \frac{B_{p-i}}{p-i}\ (\mathrm{mod}\ p),$$

which is exactly Herbrand's theorem.

**Special values in the Stickelberger relation.**

For a nontrivial character $\omega^j$, define
$$S(k,j,p)=\sum_{a=1}^{p-1}a^k\omega^j(a).$$
At $k=1$,
$$\frac{S(1,j,p)}p=B_{1,\omega^j}=-L(0,\omega^j).$$
Thus the coefficient already appearing in Herbrand's theorem is a special $L$-value:
$$e_\chi\theta=-L(0,\chi^{-1})e_\chi.$$
For $p=37$ and $\chi=\omega^5$, we use $j=31$. The factor $p$ explains why $S(1,31,37)$ has valuation $2$, while $B_{1,\omega^{31}}$ has valuation $1$.

```gp
S(k,j,p) = sum(a=1,p-1, a^k*teichmuller(a+O(p^10))^j);

{
    my(s31 = S(1,31,p));
    print("S(1,31,p) = ", s31+O(p^4));
    print("S(1,31,p)/p = ", s31/p+O(p^3));
    print("valuations: ", valuation(s31,p), " -- ", valuation(s31/p,p));
    print("Agreement with B1 modulo p^2: ",
          Mod(lift(s31/p),p^2) == B1_OmegaPow(31,p^2));
}
```

### 4. Further observation

For the computations below let $M=p^s$ and take an auxiliary prime $\ell\equiv1\pmod M$. Write $\tau_c$ for the automorphism acting as $\zeta_p\mapsto\zeta_p^c$ and trivially on $\mathbb Q(\zeta_\ell)$.

We use the normalization of the original scripts: choose $b_\ell$ by
$$
b_\ell\equiv c\pmod{p^{\mathrm{BPREC}}},\qquad b_\ell\equiv1\pmod\ell,
$$
and set
$$
\theta(\ell)=(\tau_c-b_\ell)
\sum_{\substack{a\bmod p\ell\\(a,p\ell)=1}}
\Big\{\frac{a}{p\ell}\Big\}\tau_a^{-1}
\in\mathbb Z[\operatorname{Gal}(\mathbb Q(\zeta_{p\ell})/\mathbb Q)].
$$
Here $c=delta=2$ when $p=37$; the integer generator `delta` is distinct from the derived scalar $\delta(\ell)$.

Fix a generator $\sigma_\ell$ of $G_\ell=\operatorname{Gal}(\mathbb Q(\zeta_{p\ell})/\mathbb Q(\zeta_p))$ and put
$$
D_\ell=\sum_{i=1}^{\ell-2}i\,\sigma_\ell^i,\qquad
N_\ell=\sum_{\tau\in G_\ell}\tau,\qquad
(\sigma_\ell-1)D_\ell=(\ell-1)-N_\ell.
$$
The derived Stickelberger element is the scalar $\delta(\ell)\in\mathbb Z/M\mathbb Z$ satisfying
$$
D_\ell e_\chi\theta(\ell)\equiv\delta(\ell)e_\chi N_\ell\pmod M.
$$

In this presentation $\chi=\omega^5$ and $M=37$. The code checks this congruence by checking all coefficients after $\chi$-projection. The choice above uses the ordinary integer lift of $c$, as in the original scripts. Rubin's lift on higher $p$-power roots of unity is the Teichmüller lift; the two agree modulo $p$. The hypotheses needed for the class-group bound in §5 are stronger than those of this illustrative mod-$p$ computation.

**Computing $\delta(\ell)$.** We retain the function names and normalization from the original scripts. The functions also allow a squarefree level $n$ whose prime factors are $1\pmod M$.

The computation has three steps:

1. Use the coefficient formula
   $$[\tau_a]\theta(n)=-\left\lfloor\frac{b(n)\,\operatorname{lift}(a^{-1}\bmod pn)}{pn}\right\rfloor.$$
2. Project by $e_\chi$ and apply $D_n$, keeping the $\varphi(n)$ coefficients indexed by automorphisms acting trivially on $\mathbb Q(\zeta_p)$.
3. Check that these coefficients are all equal. Since the corresponding coefficients of $e_\chi N_n$ are $1/(p-1)$, multiply their common value by $p-1$.

The auxiliary functions below implement the CRT lifts and enumerate the units modulo $n$.

```gp
delta_pn(n) =
{
    if(n == 1, return(lift(Mod(delta, p))));
    return(lift(chinese(Mod(delta, p), Mod(1, n))));
}

b(n) =
{
    if(n == 1, return(lift(Mod(delta, p^BPREC))));
    return(lift(chinese(Mod(delta, p^BPREC), Mod(1, n))));
}

UnitsModN(n) = if(n == 1, [1], select(v -> gcd(v,n) == 1, [1..n]));
```

The next function returns the coefficient vector of $D_n e_\chi\theta(n)$ in these coordinates.

```gp
RHS_DTheta_Vec(n,k,M) =
{
    my(m = p*n, s = valuation(M,p), F, beta, dm, wd, w, invp1, U, phin,
       pos, X);

    if(M != p^s || s < 1, error("M must be a positive power of p"));
    if(s > BPREC,
        error("valuation(M,p) > BPREC: raise BPREC so that b(n) = delta mod M"));
    if(n > 1,
        if(!issquarefree(n) || gcd(n,p) > 1,
            error("n must be squarefree and prime to p"));
        F = factor(n);
        for(i=1, matsize(F)[1],
            if((F[i,1]-1) % M,
                error(Str("prime ", F[i,1],
                    " | n must satisfy l = 1 mod M = ", M)))));

    beta  = b(n);
    dm    = delta_pn(n);
    wd    = TeichLiftMod(delta,p,s);
    w     = Mod(wd,M)^(-k);              \\ chi^{-1}(delta)
    invp1 = Mod(p-1,M)^(-1);
    U     = UnitsModN(n);
    phin  = #U;
    pos   = vector(n);
    for(j=1, phin, pos[U[j]] = j);

    \\ X[j] = coefficient of e_chi * Theta_Element(n) at tau_a:
    \\ (e_chi theta)_a = (p-1)^{-1} Sum_i chi^{-1}(delta)^i theta_{delta^{-i} a},
    \\ and lift((delta^{-i} a)^{-1}) = delta^i a^{-1} mod m.
    X = vector(phin);
    for(j=1, phin,
        my(a  = if(n==1, 1, lift(chinese(Mod(1,p), Mod(U[j],n)))));
        my(ti = lift(Mod(a,m)^(-1)), tot = Mod(0,M), wi = Mod(1,M));
        for(i=0, p-2,
            tot += wi * (-((beta*ti)\m));
            wi  *= w;
            ti   = (ti*dm) % m);
        X[j] = tot * invp1);

    \\ apply D_l for each prime l | n, acting through (Z/nZ)^*
    if(n > 1,
        for(t=1, matsize(F)[1],
            my(l = F[t,1], g = lift(znprimroot(l)));
            my(sl = if(n/l == 1, g % n,
                       lift(chinese(Mod(g,l), Mod(1,n/l)))));
            my(sinv = lift(Mod(sl,n)^(-1)), R = vector(phin));
            for(j=1, phin,
                my(u = U[j], acc = Mod(0,M));
                for(i=1, l-2,
                    u = (u*sinv) % n;
                    acc += i * X[pos[u]]);
                R[j] = acc);
            X = R));
    return(X);
}
```

Finally, `Delta_Fast(n, k, M)` verifies proportionality to $e_\chi N_n$ and returns $\delta(n)$ as a residue class modulo $M$. It raises an error if the coefficients are not constant.

```gp
Delta_Fast(n,k,M) =
{
    my(X = RHS_DTheta_Vec(n,k,M));
    for(j=2, #X, if(X[j] != X[1],
        error("D_n theta(n,chi) is NOT a multiple of N_n mod M")));
    return((p-1)*X[1]);
}
```

### 5. Rubin's theorems

Let $d(\ell)$ be the largest divisor of $M$ dividing $\delta(\ell)$.

**Corollary (Rubin, Cor. 4.2 in [KSGS]).** Under the hypotheses on $M$ and the auxiliary primes in [KSGS, §4] (in particular $d(1)\,\#A^\chi\,p\mid M$): Let $\ell$ be either auxiliary prime or $1$, let $B\subseteq A$ be the subgroup generated by the classes of primes dividing $\ell$. Then $\#\big(A^\chi/B^\chi\big) \mid d(\ell)$. 

In particular ($\ell=1$): 

* $\#A^\chi \mid d(1)$, and $d(1)\sim B_{1,\chi^{-1}}$. 

* Combined with the analytic class number formula this yields $\#A^\chi=p^{v_p B_{1,\chi^{-1}}}$ for all odd $\chi\ne\omega$ — in particular Ribet's converse to Herbrand.

**Example $p=37$.** $$\#A^{\omega^5}=p^{v_p B_{1,\omega^{31}}}=p^{v_p B_{32}}$$

## Part II. Cyclotomic units

We now return to $p=37$ and construct an unramified Kummer extension using a cyclotomic unit. This gives a complementary way to detect nontrivial ideal classes.

### Class groups and Bernoulli numbers

Let $\mathbb{Q}(\zeta_p)$ be the cyclotomic field for a prime $p$, and let us denote by $\mathrm{Cl}$ its class group.

Denote by $B_n$ the $n$-th Bernoulli number, defined by the generating function
$$
\sum_{n=0}^{\infty}\frac{B_n}{n!}t^n=\frac{t}{e^t-1}.
$$

Essentially, Bernoulli numbers are the values of the Riemann zeta function at negative integers, since
$$
\zeta(1-n)=-\frac{B_n}{n}\qquad(n\ge2).
$$

**Theorem (Kummer)**. $p \mid \# \mathrm{Cl}$ if and only if $p \mid B_i$ for some $i = 0, 2, \ldots, p - 3$.

Such primes are called irregular primes. If we denote by $A$ the $p$-component of $\mathrm{Cl}$, then $p$ is irregular if and only if $A \neq 0$.

But what information do invidual Bernoulli numbers give? There is some natural decomposition $A = \bigoplus\limits_{i \in \mathbb{Z}/(p-1)\mathbb{Z}} A^{(i)}$ which will be explained later, so that the following theorem holds.

**Theorem (Herbrand-Ribet)**. $A^{(i)} \neq 0$ if and only if $p \mid B_{p-i}$ for odd $i$ with $3\le i\le p-2$.

The converse direction which gives non-trivial class from divisibility of Bernoulli number is the hardest and the most essential part.

The first irregular prime is $37$.

The script below computes the class number of $K=\mathbb{Q}(\zeta_p)$.

```gp
p=37;
K = bnfinit(polcyclo(p)); /* p is small, can construct bnf */
```

```gp
print("Class number of Q(zeta_", p, ") = ", K.clgp[1]);
```

Odd Bernoulli numbers vanish, except for $B_1=-\frac{1}{2}$, so they can't have any meaningful information. The fact that $v_p(B_{p-1})=-1$ is a consequence of the von Staudt--Clausen theorem.

**Theorem(von Staudt-Clausen)**. $B_n + \sum\limits_{p - 1\mid n} \frac{1}{p} \in \mathbb{Z}$ for all positive even $n$.

### Detecting classes through a Kummer extension

How can we provide non-trivial class $[\mathfrak{a}]$ in $A$? Recall basic class field theory.

There is unique unramified abelian extension $H/K$ such that Artin symbol induces isomorphism $Cl(K) \to \mathrm{Gal}(H/K)$. It can be characterized as maximal abelian everywhere unramified extension of $K$.

Thus by Galois theory $\mathbb{Z}/p\mathbb{Z}$ **quotients** of $\mathrm{Cl}(K)$ are just $\mathbb{Z}/p\mathbb{Z}$-extensions of $K$ which are everywhere unramified.

But obviously $\zeta_p \in K$, so by Kummer theory such extensions are always of the form $K(\alpha^{1/p})$. Suppose $\alpha$ is a unit of $K$. Then the only ramification can occur above $1 - \zeta_p$, the only place in $K$ above $p$. This leads to the following

**Observation.** A unit defining a nontrivial unramified Kummer extension detects a nontrivial quotient of $\mathrm{Cl}(K)$.

Suppose we are given a unit generating unramified extension as above. How do we extract class from it?
If Artin symbol of $\mathfrak{a}$ is non-zero then the image of the class of $\mathfrak{a}$ is non-trivial in $\mathbb{Z}/p\mathbb{Z}$-quotient corresponding to the extension. By Chebotarev we may assume that the residue field of $\mathfrak{a}$ is $\mathbb{F}_\ell$ for some rational prime $\ell$.

**Proposition**. Suppose $\alpha$ is a unit in $K$, $\mathfrak{a}$ is a prime in $K$ with residue field $\mathbb{F}_\ell$ for $\ell \neq p$. Then the Frobenius of $\mathfrak{a}$ in $K(\alpha^{1/p})$ is non-trivial if and only if reduction $\overline{\alpha} \in \mathbb{F}_\ell$ is not a $p$-th power.

```gp
\\ Test at the degree-one prime fixed by zeta_p = g^((l-1)/p).
check(l, u) =
{
    my(g = znprimroot(l), zpmodl, umodl);
    zpmodl = g^((l-1)/p);
    umodl = subst(lift(u), x, zpmodl);
    return(umodl^((l-1)/p) != 1);
}
```

### The local condition at $p$

The only thing we are left to check is that $K(u^{1/p})$ is unramified at $1 - \zeta_p$. How can we do that?

If $u$ is $1-\zeta_p$-adically $p$-th root, then the local extension is obviously unramified.

So we need to check that locally at $p$ it is possible to extract the $p$-th root of $u$. It can be verified for sufficiently close approximations by Hensel's lemma.

**Hensel's Lemma.** Suppose $K$ is a field complete under non-archimedean valuation $\lvert \cdot \rvert$, $f \in \mathcal{O}_K[x]$ is a monic polynomial. If $\lvert f(x) \rvert < \lvert f^\prime(x) \rvert^2$ for some $x \in \mathcal{O}_K$ then $f$ has a root in $\mathcal{O}_K$.

Applying this lemma to $f = x^p - u$ in completion of $K$ by $1 - \zeta_p$, since $v(p) = p - 1$ we see that it is enough to check that $f$ has a root modulo $(1-\zeta)^{2p-1}$.

For the computation we use PARI/GP's [`nfislocalpower`](https://pari.math.u-bordeaux.fr/dochtml/html-stable/General_number_fields.html#nfislocalpower), which tests the $p$-th-power condition in the completion $K_{\mathfrak p}$ itself.

```gp
check_plocalpower(u) =
{
    my(pr = idealprimedec(K, p)[1]);
    return(nfislocalpower(K, pr, u, p));
}

print("u is a p-th power in K_p: ", check_plocalpower(u));
```

Finally, since in this example $p = 37$ is small enough, we can use bnf in order to verify that our function
works correctly.

```gp
get_ideal(l) =
{
    my(g = znprimroot(l), zpmodl, J);
    zpmodl = g^((l-1)/p);
    J = idealadd(K, l, Mod(x - lift(zpmodl), polcyclo(p)));
    return(J);
}

is_principal(bnf, a) = !bnfisprincipal(bnf, a, 0);

{
    print("ell -- nontrivial Frobenius -- nonprincipal ideal");
    for(t = 1, 1000,
        my(l = t*p+1);
        if(isprime(l),
            print(l, " -- ", check(l, u), " -- ", !is_principal(K, get_ideal(l)))));
}
```

### Special values of the $p$-adic $L$-function

To explain the local $p$-th-power condition for a cyclotomic unit, we use the special values of the $p$-adic $L$-function.

For a nontrivial even power $\omega^i$ and positive integers $n\equiv i\pmod{p-1}$, set
$$
L_p(1-n,\omega^i)=-(1-p^{n-1})\frac{B_n}{n}.
$$
The Kummer congruences make these values continuous on their dense set of interpolation points.

**Theorem (Kummer).** If positive integers $m,n$ satisfy
$m\equiv n\pmod{(p-1)p^N}$ and $p-1\nmid n$, then
$$
(1-p^{n-1})\frac{B_n}{n}
\equiv
(1-p^{m-1})\frac{B_m}{m}\pmod{p^N}.
$$

In fact $L_p(s,\omega^i)$ is $p$-adic analytic. We will use its values at interpolation points and at $s=1$ to study local units.

We use the formula for all non-positive integer values.

**Proposition**. For integer $n \ge 1$
$$
L_p(1-n,\omega^i) = -(1-\omega^{i-n}(p)p^{n-1})\frac{B_{n,\omega^{i-n}}}{n},
$$
where $B_{n,\chi}$ is the $n$-th generalized Bernoulli number associated with character $\chi$ of conductor $f$, defined by $\sum\limits_{n=0}^\infty \frac{B_{n,\chi}}{n!} t^n = \sum\limits_{a=1}^f\frac{\chi(a)te^{at}}{e^{ft}-1}$.

Recall that a non-trivial power of the Teichmüller character vanishes at $p$. Hence, if $n \not\equiv i \mod (p-1)$, the Euler factor $(1-\omega^{i-n}(p)p^{n-1})$ equals to 1. Otherwise, since $B_{n,1} = B_n$, proposition recovers the definition of $L_p(s,\omega^i)$.

Generalized Bernoulli numbers computed by the next formula

**Proposition**. For Dirichlet character $\chi$ of conductor $f$
$$
B_{n,\chi} = f^{n-1}\sum\limits_{a=1}^f\chi(a)B_n\left(\frac{a}{f}\right),
$$
where $B_n\left(\frac{a}{f}\right)$ is the value of $n$-th Bernoulli polynomial at $\frac{a}{f}$. 

$\sum\limits_{n=0}^\infty \frac{B_n(x)}{n!} t^n = \frac{te^{xt}}{e^t-1}$

```gp
BERN(n,i)=if(i%(p-1)!=0,p^(n-1)*sum(aa=1,p-1,teichmuller(aa+O(p^5))^i*bernpol(n,aa/p)),bernfrac(n)+O(p^5));

LFUN(m,i)=if((i-(lift(1-m+O(p^5))))%(p-1)!=0,-BERN(lift(1-m+O(p^5)),i-(lift(1-m+O(p^5))))/(lift(1-m+O(p^5))),-(1-p^(lift(1-m+O(p^5))-1))*bernfrac(lift(1-m+O(p^5)))/(lift(1-m+O(p^5)))+O(p^5));
```

$$L_p(0,\omega^i)=L(0,\omega^{i-1})
=-B_{1,\omega^{i-1}}=-\frac{S(1,i-1,p)}p.$$

### Constructing the cyclotomic unit

We now turn to the construction of the unit $u$.

Fix a primitive root $g$ modulo $p$. For instance $5$ for $p=37$.
Consider some cyclotomic unit
$$
\xi_g=\zeta_p^{\frac{1-g}{2}}\frac{1-\zeta_p^g}{1-\zeta_p}.
$$
By the choice of the factor $\zeta_p^{\frac{1-g}{2}}$, it is invariant under complex conjugation.

However, we need this unit to satisfy some additional properties. More precisely, we want it to admit a $p$-th root locally at the prime above $p$. This can be achieved by projecting the cyclotomic unit onto the appropriate isotypic component, using the relation between $\omega^i$-components of cyclotomic units and special values of $L_p(s,\omega^i)$.

Let's consider $\operatorname{pr}_{\omega^i,N}$ -- truncation modulo $p^N$ of the $\operatorname{pr}_{\omega^i} = \sum\limits_{a=1}^{p-1}\sigma_a\omega^{-i}(a)$ -- projector onto $\omega^i$-isotypical component.

If we denote by the $\omega^{i}_N(a)$ truncation of the $p$-adic number $\omega^{i}(a)$ modulo $p^N$ then the truncation of the projector can be written in the next form:
$$\operatorname{pr}_{\omega^i,N} = \sum\limits_{a=1}^{p-1}\sigma_a\omega^{-i}_N(a)$$

This operator is unnormalized: it equals $(p-1)e_{\omega^i}$. Since $p-1$ is a $p$-adic unit, it generates the same projected $\mathbb Z_p$-lattice as the normalized idempotent.

Finally, the unit $u$ described above is constructed as follows:

$$
u = \operatorname{pr}_{\omega^i,N}(\xi_g),
$$
where $i$ -- irregular index and $N$ is sufficiently large.

```gp
zetap = Mod(x, polcyclo(p));

act_galois(i, alpha) = Mod(subst(lift(alpha),x,x^i), polcyclo(p));
act_group_alg(v, alpha) = prod(i=1,p-1,act_galois(i, alpha)^v[i]);

teichmuller_projector(i, N) = vector(p-1, sigma, lift(teichmuller(sigma + O(p^N))^(-i)));

get_unit(i,j) = act_group_alg( teichmuller_projector(i, 2), zetap^((1-j)/2) * (1 - zetap^j)/(1-zetap));

print("Stored u equals the constructed cyclotomic unit: ", u == get_unit(32,5));
```

We now turn to the justification of this choice of the unit.

**One branch, several special values.**

We now follow the same branch $L_p(s,\omega^{32})$ through several arguments. The value at $s=0$ connects to Stickelberger, while
$$L_p(-31,\omega^{32})=-(1-37^{31})\frac{B_{32}}{32}.$$
The arguments $1-p^r$ converge $p$-adically to $1$, the point relevant to the local index of cyclotomic units.

We compare this branch with the regular branch $\omega^2$. In the displayed computations, the first branch has valuation $1$, while the second has residue $3$ modulo $37$. This illustrates the congruence used below:
$$L_p(0,\omega^{32})\equiv L_p(1,\omega^{32})\equiv0\pmod{37}.$$

### Local index and the value at $s=1$

The following is the key theorem for it.

**Definition**. The group of cyclotomic units $C$ is the subgroup of units generated by $(1-\zeta_p^a)/(1-\zeta_p)$.

Denote by $\pi$ unique principal prime ideal $(1-\zeta_p)$ lying above $p$.
Let $U_{1,\pi}$ be the group of principal $\pi$-local units of $K$ and $C_{1,\pi} = (C \otimes \mathbb{Z}_p) \cap U_{1,\pi}$.

Note that for sufficciently large $N$ unit $u$ generates corresponding component $C_{1,\pi}^{(i)} = \mathrm{pr}_{\omega^i} C_{1,\pi}$.

**Theorem**.
$$
v_p\left(\left[U_{1,\pi}^{(i)} : C_{1,\pi}^{(i)} \right]\right) = v_p\left(L_p(1,\omega^i)\right)
$$

This theorem can be proven by analyzing the following formula for value of $L_p(s,\omega^i)$ at $s=1$:

**Theorem**.
$$
L_{p}(1,\omega^{i})=-\left(1-\frac{1}{p}\right)\frac{\tau(\omega^{i})}{p}\sum\limits^{p}_{a=1}\omega^{-i}(a)\log_{p}(1-\zeta^a)
$$

Thus it remains to check that the corresponding value at $s=1$ is divisible by $p$. This follows from the following congruence.

**Theorem**. Let $\omega^i$ be a non-trivial power of the Teichmüller character. Then, for any $m,n \in \mathbb{Z}$, one has
$$
L_p(m,\omega^i) \equiv L_p(n,\omega^i) \pmod p.
$$

**Corollary**. Suppose $i$ is even. Then $p \mid B_{i}$ if and only if $p \mid L_p(1, \omega^i)$.

We know that the value of this $p$-adic $L$-function at the interpolation point corresponding to an irregular index is divisible by $p$, since the corresponding Bernoulli number is divisible by $p$. Hence its value at $s=1$ is also divisible by $p$.

Thus divisibility of the Bernoulli number gives the divisibility of the local index needed in the construction.

The remaining issue is global. A unit which is locally a $p$-th power defines a nontrivial Kummer extension only if it is not a $p$-th power in $K$ itself.

Let $E=\mathcal O_K^\times$. Divisibility of
$[U_{1,\pi}^{(i)}:C_{1,\pi}^{(i)}]$ explains the local condition. To rule out a global $p$-th root for a generator of the cyclotomic-unit component, it suffices to know
$$p\nmid[E_{1,\pi}^{(i)}:C_{1,\pi}^{(i)}].$$
The Kummer–Vandiver conjecture provides this condition. In the example $p=37$, the residue test at $\ell=149$ has already directly shown that $u$ is not a global $p$-th power.

### The role of the Kummer–Vandiver conjecture

We have already seen that the Kummer extension constructed from $u$ is non-trivial. The Kummer--Vandiver conjecture predicts that this is the general situation.

**Kummer--Vandiver conjecture (first form)**. The prime $p$ does not divide the class number of the maximal real subfield $K^+$ of $K=\mathbb{Q}(\zeta_p)$.

By the analytic class number formula, this statement is equivalent to the following one.

**Kummer--Vandiver conjecture (second form)**. The prime $p$ does not divide the index of the group of cyclotomic units in the full unit group modulo roots of unity of $K$.

Cyclotomic units of $K$ genereated by $\frac{1-\zeta_p^a}{1-\zeta}$ for $a=2,\cdots,\frac{p-1}{2}+1$.

The following script verifies the Kummer--Vandiver conjecture for primes small enough that a system of fundamental units of $K$ can be computed.

```gp
z = Mod(x, polcyclo(p));
Cunits = vector((p-1)/2, t, (1-z^(t+1))/(1-z));

{
    my(unit_rank = (p-1)/2-1, unit_exponents = Mat(), coord);
    for(t = 1, #Cunits,
        coord = bnfisunit(K, Cunits[t]);
        unit_exponents = concat(unit_exponents, vector(unit_rank, j, coord[j])~));
    my(unit_lattice = mathnf(unit_exponents));
    print("Index of cyclotomic units modulo roots of unity = ", abs(matdet(unit_lattice)));
}
```

It is enough to know that $p$ does not divide
$[E_{1,\pi}^{(i)}:C_{1,\pi}^{(i)}]$. A generator of the cyclotomic-unit component then cannot be a global $p$-th power. In our example, this is also witnessed directly by reduction at a prime above $\ell=149$.

## Part III. Kummer duality and reflection

Why does a cyclotomic unit with character $\omega^{32}$ detect an ideal class with character $\omega^5$? Take the prime $\mathfrak l$ above $149$ used in Part II. We already know that its Frobenius acts nontrivially on $K(\sqrt[37]{u})$. Its action on a chosen root gives a more precise answer.

### The Kummer symbol at the prime above 149

Write $\beta^p=u$ and use **arithmetic Frobenius**, which acts on the residue field by $x\mapsto x^\ell$. At a prime $\mathfrak l$ of residue degree one with $\ell\ne p$, the unit $\beta$ satisfies
$$
\langle [\mathfrak l],[u]\rangle
=\frac{\operatorname{Frob}_{\mathfrak l}(\beta)}{\beta}\in\mu_p,
\qquad
\overline{\langle [\mathfrak l],[u]\rangle}
=\bar u^{(\ell-1)/p}.
$$
Reduction embeds $\mu_p$ in $\mathbf F_\ell^\times$. As in `get_ideal`, choose $\mathfrak l$ by sending $\zeta_p$ to $g^{(\ell-1)/p}$ for the same primitive root $g$ modulo $\ell$. Taking a discrete logarithm to this base recovers the exponent of $\zeta_p$ in the symbol.

```gp
{
    my(ell = 149, g = znprimroot(ell));
    my(zbar = g^((ell-1)/p), ubar = subst(lift(u), x, zbar));
    my(symbol = ubar^((ell-1)/p), exponent = znlog(symbol, zbar, p));
    print("zeta mod l = ", lift(zbar), ", u mod l = ", lift(ubar));
    print("u^((ell-1)/p) mod l = ", lift(symbol));
    print("Kummer pairing = zeta_", p, "^", exponent);
}
```

### Why the characters add to one

The computation gives $\langle[\mathfrak l],[u]\rangle=\zeta_{37}^{23}$. To identify the two character spaces that it pairs, let $H/K$ be the maximal unramified abelian extension of exponent $p$, where $K=\mathbf Q(\zeta_p)$. Write $A=\mathrm{Cl}(K)[p^\infty]$ and define its unramified Kummer radical by
$$
R=\{[\alpha]\in K^\times/K^{\times p}:K(\sqrt[p]{\alpha})\subseteq H\}.
$$
Artin reciprocity identifies $A/pA$ with $\operatorname{Gal}(H/K)$. Kummer duality gives a perfect pairing
$$
(A/pA)\times R\longrightarrow\mu_p,\qquad
(c,[\alpha])\longmapsto \frac{c(\sqrt[p]{\alpha})}{\sqrt[p]{\alpha}}.
$$
For $\sigma\in\Delta$, equivariance gives $\langle\sigma c,\sigma[\alpha]\rangle=\sigma\langle c,[\alpha]\rangle$. If $c$ has character $\omega^j$ and $[\alpha]$ has character $\omega^i$, bilinearity raises the pairing to $\omega^{i+j}(\sigma)$ on the left. On the right $\sigma$ acts on $\mu_p$ by $\omega(\sigma)$. A nontrivial pairing therefore requires $i+j\equiv1\pmod{p-1}$, and perfectness identifies $R^{(1-j)}$ with $\operatorname{Hom}((A/pA)^{(j)},\mu_p)$.

For $\chi=\omega^5$ the three characters in Parts I and II now fit together.

| Object | Character at $p=37$ |
| :--- | :--- |
| Ideal class $[\mathfrak l]$ | $\chi=\omega^5$ |
| Stickelberger coefficient $B_{1,\chi^{-1}}$ | $\chi^{-1}=\omega^{31}$ |
| Unit class $[u]$ in the radical | $\omega\chi^{-1}=\omega^{32}$ |

The interpolation formula gives the same shift analytically: $L_p(0,\omega^{32})=-B_{1,\omega^{31}}$.

### Washington's reflection theorem

Let $r_k=\dim_{\mathbf F_p}(A/pA)^{(k)}$, with indices taken modulo $p-1$.

**Theorem 10.9 (Washington).** For an odd prime $p$, an even index $i$ and an odd index $j$ with $i+j\equiv1\pmod{p-1}$,
$$r_i\le r_j\le r_i+1.$$

Where does the extra dimension come from? For $[\alpha]\in R$, unramifiedness forces every finite valuation of $\alpha$ to be divisible by $p$. Thus $(\alpha)=\mathfrak a^p$, and the assignment
$$
\Phi:R\longrightarrow A[p],\qquad [\alpha]\longmapsto[\mathfrak a]
$$
is well defined and $\Delta$-equivariant. Its kernel consists of the radical classes represented by global units. Write $E=\mathcal O_K^\times$. Since $\dim R^{(i)}=r_j$ by duality and $\dim A[p]^{(i)}=r_i$, we obtain
$$
r_j\le r_i+\dim_{\mathbf F_p}(E/E^p)^{(i)}.
$$
For nonzero even $i$, this unit space has dimension one. For odd $j\ne1$, the corresponding unit space is zero: the only odd contribution to $E/E^p$ comes from $\mu_p$ in character $\omega$. Applying the same bound with $i,j$ interchanged gives $r_i\le r_j$. For the pair $(0,1)$, the norm to $\mathbf Q$ gives $r_0=0$, and $(E/E^p)^{(0)}=0$ gives $r_1=0$ by the displayed bound.

### The reflected pair $(32,5)$

Part II computed the index of the cyclotomic units modulo roots of unity as $1$. By the cyclotomic-unit index formula, Vandiver's condition holds at $37$, so $A^{(32)}=0$. Reflection gives $r_5\le1$. The pairing $\langle[\mathfrak l],[u]\rangle=\zeta_{37}^{23}$ gives $r_5\ge1$. Hence
$$
r_{32}=0,\qquad r_5=1,\qquad R^{(32)}=\langle [u]\rangle_{\mathbf F_{37}},
$$
where the last expression denotes the line spanned by the class of $u$. This unit represents the extra dimension in the theorem: its ideal is trivial, so $[u]\in\operatorname{Ker}\Phi$, while its pairing with $[\mathfrak l]$ is nontrivial.

We can check the ranks independently using the class group from Part II. It is cyclic of order $37$, and $J=\mathfrak l$ represents a generator. The ratio of the class coordinates of $\sigma_2 J$ and $J$ gives the eigenvalue of $\sigma_2$ on $A/pA$. Since $2$ generates $\mathbf F_{37}^\times$, its discrete logarithm gives the character exponent.

```gp
{
    if(K.cyc != [p], error("This check uses a cyclic class group of order p"));
    my(c = Mod(bnfisprincipal(K, J, 0)[1], p));
    my(sc = Mod(bnfisprincipal(K, jJ, 0)[1], p));
    if(c == 0, error("The chosen ideal represents the trivial class"));
    my(eigenvalue = sc/c, character = znlog(eigenvalue, Mod(2,p), p-1));
    print("Class group invariants = ", K.cyc);
    print("sigma_2 eigenvalue = ", lift(eigenvalue), ", character exponent = ", character);
    print("r_32 = ", character == 32, ", r_5 = ", character == 5);
}
```

Part I supplies the order of the group that the symbol detects: $v_{37}(B_{1,\omega^{31}})=1$ gives $\#A^{(5)}=37$. Thus $[\mathfrak l]$ generates $A^{(5)}$, and adjoining a $37$th root of $u$ realizes its unramified extension of degree $37$.

**Consequence under Vandiver.** If $p\nmid h(K^+)$, all even $r_i$ vanish. Theorem 10.9 then gives $r_j\le1$ for each odd $j$, so each odd component $A^{(j)}$ is cyclic. This is the base-field cyclicity statement behind Washington's Theorem 10.14, which describes the class group as a module over the Galois group ring in the cyclotomic tower under the same hypothesis.
